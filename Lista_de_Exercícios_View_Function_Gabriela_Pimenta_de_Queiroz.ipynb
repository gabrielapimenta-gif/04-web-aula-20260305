**1. Explique com suas palavras por que o uso de funções torna o código mais modular, reutiliźavel e fácil de entender.**

As funções são blocos de código que realizam uma tarefa específica e podem ser chamados a qualquer momento no programa. O uso delas traz várias vantagens:

1.  **Modularidade:**
    *   **O que é:** Imagine que seu programa é um grande quebra-cabeça. Sem funções, todas as peças estariam misturadas. Com funções, você organiza o código em "módulos" ou "blocos" independentes, onde cada função é responsável por uma parte específica da lógica do programa. Por exemplo, uma função pode calcular uma média, outra pode exibir uma mensagem na tela e outra pode ler dados de um arquivo.
    *   **Benefício:** Isso ajuda a gerenciar a complexidade de programas grandes. Em vez de ter um código enorme e confuso, você tem várias partes menores e bem definidas. Se algo der errado, é mais fácil identificar qual módulo (função) está com problema, o que facilita a depuração.

2.  **Reutilização:**
    *   **O que é:** Pense em uma receita de bolo. Se você precisa fazer o mesmo tipo de bolo várias vezes, você não reescreve a receita do zero toda vez. Você simplesmente a segue novamente. Funções funcionam da mesma forma. Se você tem uma tarefa que precisa ser realizada em diferentes pontos do seu programa (ou até mesmo em outros programas), você a encapsula em uma função.
    *   **Benefício:** Em vez de copiar e colar o mesmo código repetidamente (o que é uma má prática e pode levar a erros e dificuldades de manutenção), você escreve a lógica uma única vez dentro de uma função e a chama sempre que precisar. Isso economiza tempo, reduz o tamanho do código e garante que, se houver uma correção ou melhoria, ela precise ser feita apenas em um lugar.

3.  **Facilidade de Entendimento (Legibilidade):**
    *   **O que é:** Um programa sem funções pode ser como um livro sem capítulos, parágrafos ou pontuação – uma parede de texto difícil de ler e entender. Funções agem como esses capítulos e parágrafos, dividindo o código em unidades lógicas com nomes descritivos.
    *   **Benefício:** Ao dar nomes significativos às funções (por exemplo, `calcular_area()`, `enviar_email()`, `validar_entrada()`), você torna o propósito de cada bloco de código muito mais claro. Isso permite que você (ou outra pessoa que leia seu código) entenda rapidamente o que o programa faz em um nível mais alto, sem precisar se aprofundar imediatamente nos detalhes de cada implementação. Torna o código mais limpo, autoexplicativo e, consequentemente, mais fácil de manter e colaborar.

**5. Crie uma função chamada IMC que receba dois argumentos: peso e altura. Ela deve
retornar o  índice de Massa Corporal (IMC = peso / altura2). Depois, crie uma função chamada ClassificaIMC que chame a função IMC e retorne uma classificação (“abaixo do peso”, “normal", “sobrepeso”, etc.).
imc = peso(quilos) altura2(metros)**
* Baixo peso muito grave = abaixo de 16 kg/m2
*   Baixo peso grave = entre 16 e 16,99 kg/m2
*   Baixo peso = entre 17 e 18,49 kg/m2
*   Peso normal = entre 18,50 e 24,99 kg/m2
* Sobrepeso = entre 25 e 29,99 kg/m2
* Obesidade grau I = entre 30 e 34,99 kg/m2
* Obesidade grau II = entre 35 e 39,99 kg/m2
* Obesidade grau III (obesidade mórbida) = maior que 40 kg/m2

````python
def IMC(peso, altura):
    if altura == 0:
        raise ValueError("A altura não pode ser zero.")
    return peso / (altura ** 2)

def ClassificaIMC(peso, altura):
    indice_imc = IMC(peso, altura)

    if indice_imc < 16:
        return "Baixo peso muito grave"
    elif 16 <= indice_imc < 16.99:
        return "Baixo peso grave"
    elif 17 <= indice_imc < 18.49:
        return "Baixo peso"
    elif 18.50 <= indice_imc < 24.99:
        return "Peso normal"
    elif 25 <= indice_imc < 29.99:
        return "Sobrepeso"
    elif 30 <= indice_imc < 34.99:
        return "Obesidade grau I"
    elif 35 <= indice_imc < 39.99:
        return "Obesidade grau II"
    else:
        return "Obesidade grau III (obesidade mórbida)"

peso1 = 70.0
altura1 = 1.75
imc1 = IMC(peso1, altura1)
classificacao1 = ClassificaIMC(peso1, altura1)
print(f"Para peso {peso1}kg e altura {altura1}m: IMC = {imc1:.2f}, Classificação: {classificacao1}")
````

Exemplo de uso:

**9. Diferencie os conceitos de argumento por cópia e argumento por referência. Dê um exemplo em pseudocódigo que mostre a diferença entre os dois.**

Diferença entre Argumento por Cópia (Passagem por Valor) e Argumento por Referência (Passagem por Referência)

  1. Argumento por Cópia (ou Passagem por Valor)
     - Quando um argumento é passado por cópia, uma DUPLICATA (cópia) do valor da variável original é enviada para a função.
     - A função trabalha com essa cópia, e qualquer alteração feita dentro da função afeta APENAS a cópia, não a variável original fora da função.
     - Tipos de dados "imutáveis" (como números inteiros, floats, strings, tuplas em Python) geralmente se comportam como passagem por valor.

  2. Argumento por Referência (ou Passagem por Referência)
     - Quando um argumento é passado por referência, a função recebe um "ponteiro" ou uma "referência" para a localização de memória da variável original.
     - A função, portanto, trabalha DIRETAMENTE com a variável original. Qualquer alteração feita dentro da função afetará a variável original fora da função.
     - Tipos de dados "mutáveis" (como listas, dicionários, conjuntos e objetos personalizados em Python) se comportam de forma que a função pode modificar o objeto original.

  --- Exemplo em Pseudocódigo ---

  Exemplo de Passagem por Cópia
  Pseudocódigo
  FUNÇÃO INCREMENTA_COPIA(numero):
     numero = numero + 1
     IMPRIMIR("Dentro da função (cópia): ", numero)
  FIM FUNÇÃO

  VARIAVEL meu_numero = 10
  IMPRIMIR("Antes da função (original): ", meu_numero)
  CHAMAR INCREMENTA_COPIA(meu_numero)
  IMPRIMIR("Depois da função (original): ", meu_numero)

  Saída esperada:
  Antes da função (original): 10
  Dentro da função (cópia): 11
  Depois da função (original): 10  (Não foi alterado)

  Exemplo de Passagem por Referência (com objeto mutável)
  Pseudocódigo
  FUNÇÃO ADICIONA_ELEMENTO(lista):
     lista.ADICIONAR(4)
     IMPRIMIR("Dentro da função (referência): ", lista)
  FIM FUNÇÃO

  VARIAVEL minha_lista = [1, 2, 3]
  IMPRIMIR("Antes da função (original): ", minha_lista)
  CHAMAR ADICIONA_ELEMENTO(minha_lista)
  IMPRIMIR("Depois da função (original): ", minha_lista)

  Saída esperada:
  Antes da função (original): [1, 2, 3]
  Dentro da função (referência): [1, 2, 3, 4]
Depois da função (original): [1, 2, 3, 4]  (Foi alterado)

Em Python, o modelo é frequentemente chamado de "passagem por atribuição de objeto" (pass by object reference).
Isso significa que uma função recebe uma referência para o objeto. Se o objeto é mutável, ele pode ser modificado.
Se o objeto é imutável, qualquer "modificação" dentro da função na verdade cria um novo objeto local, sem afetar o original.


**10. Escreva uma função chamada Fatorial(n) que recebe como argumento um número inteiro positivo n e retorna o valor do seu fatorial. O fatorial de um n ́umero n (representado por n!) é definido como:
n! = n × (n − 1) × (n − 2) × . . . × 1**

**Por definição, 0! = 1.
Exemplo de uso:
Fatorial(5) → 120 (pois 5! = 5 × 4 × 3 × 2 × 1)**

````python
def Fatorial(n):
    if not isinstance(n, int) or n < 0:
        raise ValueError("O argumento deve ser um número inteiro não negativo.")

    if n == 0:
        return 1
    else:
        resultado = 1
        for i in range(1, n + 1):
            resultado *= i
        return resultado

print(f"Fatorial(5) -> {Fatorial(5)}")
````

**11. Crie uma função recursiva chamada Fibonacci(n) que calcula o n-ésimo termo da
sequência de Fibonacci. Lembre-se de que:**

F(0) = 0, F(1) = 1, F(n) = F(n − 1) + F(n − 2), para n > 1

**Exemplo de uso:
Fibonacci(6) → 8**

````python
def Fibonacci(n):
    if n < 0:
        raise ValueError("O índice não pode ser negativo.")
    elif n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return Fibonacci(n - 1) + Fibonacci(n - 2)

termo = 6
resultado_fibonacci = Fibonacci(termo)
print(f"Fibonacci({termo}) -> {resultado_fibonacci}")
````

Mais especificamente, uma função de primeira classe pode ser:

*   Atribuída a uma variável: Você pode guardar uma função em uma variável e usá-la através dessa variável.
*   Passada como argumento para outra função: Uma função pode ser "enviada" para outra função que a executará ou a manipulará.
*  Retornada como resultado de outra função: Uma função pode gerar e retornar uma nova função.

**17. Descreva, com suas palavras, as principais vantagens do paradigma funcional em comparação ao paradigma imperativo. Foque especialmente em aspectos como: imutabilidade, ausência de efeitos colaterais, recursão, composição de funções e legibilidade.**

O **paradigma funcional** é uma forma de construir software onde as computações são tratadas como a avaliação de funções matemáticas e onde a mudança de estado e dados mutáveis são evitados. Em contraste, o **paradigma imperativo** foca em como um programa opera, descrevendo os passos e o controle do fluxo para mudar o estado do programa.

As principais vantagens do paradigma funcional incluem:

1.  **Imutabilidade:**
    *   **O que é:** No funcional, os dados são **imutáveis**, o que significa que uma vez criados, eles não podem ser alterados. Em vez de modificar um dado existente, você cria uma nova versão modificada. (Ex: em vez de `lista.append(x)`, você faria `nova_lista = lista + [x]`).
    *   **Vantagem:** Isso simplifica drasticamente a lógica do programa, pois você nunca precisa se preocupar se outra parte do código modificou um dado que você está usando. Torna o código mais previsível e mais fácil de depurar, especialmente em sistemas concorrentes ou paralelos, onde a modificação compartilhada de estado pode levar a erros complexos.

2.  **Ausência de Efeitos Colaterais:**
    *   **O que é:** Uma função funcional idealmente só produz um resultado com base em seus argumentos de entrada, sem alterar qualquer estado fora de seu escopo (como variáveis globais, arquivos, ou a interface do usuário). Funções que evitam isso são chamadas de **puras**.
    *   **Vantagem:** Funções puras são mais fáceis de testar (dado o mesmo input, sempre o mesmo output), mais fáceis de raciocinar sobre (o que elas fazem é evidente apenas pelos seus inputs e outputs), e podem ser executadas em qualquer ordem ou em paralelo sem risco de interferências mútuas.

3.  **Recursão:**
    *   **O que é:** Enquanto o imperativo usa loops (for, while) para iterações, o funcional frequentemente prefere a **recursão**, onde uma função chama a si mesma para resolver problemas que podem ser divididos em subproblemas menores da mesma natureza.
    *   **Vantagem:** Para muitos problemas (como travessia de estruturas de dados ou cálculos sequenciais), a recursão pode levar a soluções mais elegantes, concisas e que se alinham mais naturalmente com a definição matemática do problema, tornando o código mais legível e menos propenso a erros de lógica de loop.

4.  **Composição de Funções:**
    *   **O que é:** A composição de funções é o ato de combinar funções mais simples para construir funções mais complexas. O resultado de uma função é a entrada da próxima.
    *   **Vantagem:** Permite construir programas a partir de pequenos blocos de construção testados e reutilizáveis. Isso leva a um código mais modular, onde cada função tem uma responsabilidade clara e única, facilitando a manutenção, a extensibilidade e a reusabilidade. Conceitos como _pipes_ ou _chained calls_ (`f(g(h(x)))`) são exemplos disso.

5.  **Legibilidade e Manutenibilidade:**
    *   **O que é:** Devido aos pontos anteriores (imutabilidade, ausência de efeitos colaterais, foco em funções puras), o código funcional tende a ser mais fácil de ler e entender.
    *   **Vantagem:** Uma vez que as funções se comportam como funções matemáticas (previsíveis e auto-contidas), é mais fácil compreender o que cada parte do sistema faz sem ter que rastrear um estado mutável complexo por todo o programa. Isso reduz a carga cognitiva para os desenvolvedores, tornando o código mais fácil de depurar, manter e estender.

**18. Crie uma função chamada MaiorElemento(lista) que recebe uma lista de números e retorna o maior valor encontrado. Implemente tanto a versão recursiva quanto a versão iterativa da função.**

````python
def MaiorElementoIterativo(lista):
    if not lista:
        return None

    maior_valor = lista[0]
    for elemento in lista[1:]:
        if elemento > maior_valor:
            maior_valor = elemento
    return maior_valor

def MaiorElementoRecursivo(lista):
    if not lista:
        return None
    elif len(lista) == 1:
        return lista[0]
    else:
        maior_do_resto = MaiorElementoRecursivo(lista[1:])
        if lista[0] > maior_do_resto:
            return lista[0]
        else:
            return maior_do_resto

lista1 = [3, 1, 4, 1, 5, 9, 2, 6]
print(f"Maior elemento iterativo({lista1}) -> {MaiorElementoIterativo(lista1)}")
print(f"Maior elemento recursivo({lista1}) -> {MaiorElementoRecursivo(lista1)}")
````

````python
def ComparaListas(l1, l2):
    if len(l1) != len(l2):
        return False

    for i in range(len(l1)):
        if l1[i] != l2[i]:
            return False
    return True

lista_a = [1, 2, 3]
lista_b = [1, 2, 3]

print(f"ComparaListas({lista_a}, {lista_b}) -> {ComparaListas(lista_a, lista_b)}")
````

## Por que essa prática favorece a abstração, reutilização e legibilidade do código?

1.  **Abstração:**
    *   **O que é:** Abstração é o processo de esconder os detalhes complexos de uma implementação e mostrar apenas a funcionalidade essencial. No exemplo acima, a função `ProcessarPedido` não precisa saber *como* o total é calculado ou *como* o e-mail é enviado. Ela apenas chama `CalcularTotalPedido` e `EnviarEmailConfirmacao`, tratando-os como caixas pretas que realizam suas tarefas específicas.
    *   **Benefício:** Isso simplifica a lógica da função principal. O desenvolvedor que trabalha com `ProcessarPedido` pode focar no fluxo geral do processo sem se perder nos pormenores de cada etapa. Isso torna o código mais fácil de entender em um nível mais alto.

2.  **Reutilização:**
    *   **O que é:** Reutilização significa que um pedaço de código pode ser usado em diferentes partes do programa ou até mesmo em outros projetos, sem a necessidade de reescrevê-lo.
    *   **Benefício:** As funções `CalcularTotalPedido` e `EnviarEmailConfirmacao` são unidades independentes. Se, por exemplo, precisarmos calcular o total de um carrinho de compras *antes* de finalizar o pedido, podemos chamar `CalcularTotalPedido` diretamente. Da mesma forma, `EnviarEmailConfirmacao` poderia ser usada para enviar e-mails de recuperação de senha ou promoções, sem precisar reescrever a lógica de envio de e-mail a cada vez. Isso economiza tempo e reduz a chance de erros.

3.  **Legibilidade:**
    *   **O que é:** Legibilidade refere-se à facilidade com que o código pode ser lido e compreendido por outros desenvolvedores (e por você mesmo no futuro).
    *   **Benefício:** A função `ProcessarPedido` é quase como uma lista de etapas a serem seguidas: "imprimir início, calcular total, enviar e-mail, verificar sucesso, imprimir resultado". Os nomes claros das funções auxiliares tornam o fluxo de trabalho evidente. Em vez de ter centenas de linhas de código em uma única função, dividi-las em funções menores e com responsabilidades únicas facilita a leitura, a manutenção e a identificação de problemas.
